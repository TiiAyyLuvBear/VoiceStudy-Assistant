```python
project_root = Path.cwd().parent

metadata_dir = (
    project_root
    / "data"
    / "metadata"
)

processed_audio_dir = (
    project_root
    / "data"
    / "processed"
    / "v1"
    / "audio"
)

processed_audio_dir.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
sys.path.append(str(project_root))
from src.audio.preprocessing import process_file
```

```python
metadata_files = [
    "svm_closed_set_enrollment.csv",
    "svm_closed_set_train.csv",
    "svm_closed_set_validation.csv",
    "svm_closed_set_test.csv",
    "cosine_validation_enrollment.csv",
    "cosine_validation_query.csv",
    "cosine_validation_unknown.csv",
    "cosine_test_enrollment.csv",
    "cosine_test_query.csv",
    "cosine_test_unknown.csv"
]
```

```python
metadata = {}

for file in metadata_files:
    metadata[file] = pd.read_csv(
        metadata_dir / file
    )

print(f"Loaded {len(metadata)} metadata files.")
```

```python
all_audio = pd.concat(
    metadata.values(),
    ignore_index=True
)

print("Before duplicate removal:", len(all_audio))
```

```python
all_audio = (
    all_audio
    .drop_duplicates(subset="audio_id")
    .reset_index(drop=True)
)

print("After duplicate removal:", len(all_audio))
```

```python
all_audio["processed_audio_path"] = (
    "processed/v1/audio/"
    + all_audio["audio_path"]
)
```

```python
display(
    all_audio[
        [
            "audio_id",
            "audio_path",
            "processed_audio_path"
        ]
    ].head()
)
```

```python
preprocessing_log = []
```

```python
for _, row in tqdm(
    all_audio.iterrows(),
    total=len(all_audio)
):

    input_path = (
        project_root
        / "data"
        / "audio"
        / row.audio_path
    )

    output_path = (
        processed_audio_dir
        / row.audio_path
    )

    success = False

    try:
        process_file(
            str(input_path),
            str(output_path)
        )

        success = True

    except Exception:
        success = False

    preprocessing_log.append({
        "audio_id":
            row.audio_id,
        "audio_path":
            row.audio_path,
        "processed_audio_path":
            row.processed_audio_path,
        "success":
            success
    })
```

```python
preprocessing_log = pd.DataFrame(
    preprocessing_log
)
preprocessing_log.head()
```

```python
preprocessing_log.to_csv(
    metadata_dir
    / "preprocessing_log.csv",

    index=False,
    encoding="utf-8-sig"
)
print("Saved preprocessing_log.csv")
```

```python
processed_lookup = (
    preprocessing_log[
        [
            "audio_id",
            "processed_audio_path"
        ]
    ]
)
```

### Update every metadata csv

```python
for file in metadata_files:
    df = metadata[file]
    df = df.merge(
        processed_lookup,
        on="audio_id",
        how="left"
    )

    metadata[file] = df
```

### Save updated metadata

```python
for file in metadata_files:
    metadata[file].to_csv(
        metadata_dir / file,
        index=False,
        encoding="utf-8-sig"
    )

print("Updated metadata files.")
```

### Random quality check

```python
quality_check = (

    preprocessing_log[
        preprocessing_log["success"]
    ]

    .sample(
        n=min(
            20,
            len(
                preprocessing_log[
                    preprocessing_log["success"]
                ]
            )
        ),
        random_state=42
    )

    .copy()
)
```

### Add checking columns

```python
quality_check["checked"] = False
quality_check["comment"] = ""
```

### Save quality check

```python
quality_check.to_csv(
    metadata_dir
    / "preprocessing_quality_check.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Saved preprocessing_quality_check.csv")
```

### Statistics

```python
print("========== SUMMARY ==========")
print("Total audio:")
print(len(preprocessing_log))
print("Successful:")
print(
    preprocessing_log["success"]
    .sum()
)
print("Failed:")
print(
    len(preprocessing_log)
    -
    preprocessing_log["success"].sum()
)
print()
print(
    all_audio["audio_path"]
    .str.split("/")
    .str[0]
    .value_counts()
)
```

# Kiểm tra preprocessing: Nghe ngẫu nhiên audio trước và sau xử lý; bảo đảm không mất đầu câu, cuối câu hoặc méo tiếng

### Chọn ngẫu nhiên các audio đã preprocessing thành công

```python
import pandas as pd

RANDOM_SEED = 42
NUM_SAMPLES = 20

quality_check = (
    preprocessing_log[
        preprocessing_log["success"]
    ]
    .sample(
        n=min(NUM_SAMPLES, len(preprocessing_log)),
        random_state=RANDOM_SEED
    )
    .copy()
    .reset_index(drop=True)
)

print(len(quality_check))
quality_check.head()
```

### Thêm đường dẫn audio gốc

```python
quality_check = quality_check.merge(
    all_audio[
        [
            "audio_id",
            "transcript"
        ]
    ],
    on="audio_id",
    how="left"
)
```

### Tạo cột đánh giá

```python
quality_check["head_ok"] = ""
quality_check["tail_ok"] = ""
quality_check["distortion"] = ""
quality_check["overall"] = ""
quality_check["comment"] = ""
```

```python
quality_check = quality_check[
    [
        "audio_id",
        "transcript",
        "audio_path",
        "processed_audio_path",
        "head_ok",
        "tail_ok",
        "distortion",
        "overall",
        "comment"
    ]
]
```

```python
output_path = (
    metadata_dir /
    "preprocessing_quality_check.csv"
)

quality_check.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

print(f"Saved: {output_path}")
```
